```python
class Chocolate:
    def __init__(self, price: float, additive: str = None):
        if price <= 0:
            raise ValueError("Цена должна быть больше нуля")
        self.price = price
        self.additive = additive

    def show_my_chocolate(self):
        if self.additive:
            print(f'Шоколадный батончик, добавка: {self.additive}')
        else:
            print('Обычный шоколадный батончик')

    def __str__(self):
        return f'Шоколадный баночик, {self.price} рублей'


choco1 = Chocolate(35.0)
choco2 = Chocolate(99.9, "вафли")

choco1.show_my_chocolate()
choco2.show_my_chocolate()

print(choco1)
print(choco2)


bad_chocolate = Chocolate(-10) # ошибка
```
